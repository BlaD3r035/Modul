## 1. Validación de entrada — equivalence partitioning (RF-04, RF-05, RF-06, RF-07)

`verify.verifyNumType(n, base)` devuelve `True` si `n` es un número válido en `base`.

- P1 válida: solo símbolos permitidos por la base.
- P2 inválida: un símbolo que no pertenece a la base (RF-05).
- P3 inválida: cadena vacía (RF-04).
- RF-06: letras A-F en mayúscula o minúscula. RF-07: ceros a la izquierda.

```python
class ValidacionCajaNegra(unittest.TestCase):

    def test_cp01_binario_valido(self):
        """CP-01 (P1, RF-05): '1010' es válido en binario"""
        self.assertTrue(verify.verifyNumType('1010', 2))

    def test_cp02_octal_valido(self):
        """CP-02 (P1, RF-05): '707' es válido en octal"""
        self.assertTrue(verify.verifyNumType('707', 8))

    def test_cp03_hex_mayuscula(self):
        """CP-03 (P1, RF-06): 'FF' es válido en hexadecimal"""
        self.assertTrue(verify.verifyNumType('FF', 16))

    def test_cp04_simbolo_invalido_binario(self):
        """CP-04 (P2, RF-05): '2' no es válido en binario"""
        self.assertFalse(verify.verifyNumType('2', 2))

    def test_cp05_simbolo_invalido_octal(self):
        """CP-05 (P2, RF-05): '8' no es válido en octal"""
        self.assertFalse(verify.verifyNumType('8', 8))

    def test_cp06_simbolo_invalido_hex(self):
        """CP-06 (P2, RF-05): 'G' no es válido en hexadecimal"""
        self.assertFalse(verify.verifyNumType('G', 16))

    def test_cp07_numero_vacio(self):
        """CP-07 (P3, RF-04): la cadena vacía se rechaza"""
        self.assertFalse(verify.verifyNumType('', 10))

    def test_cp08_hex_minuscula(self):
        """CP-08 (RF-06): 'ff' en minúscula es válido en hexadecimal"""
        self.assertTrue(verify.verifyNumType('ff', 16))

    def test_cp09_ceros_a_la_izquierda(self):
        """CP-09 (RF-07): '0010' es un binario válido"""
        self.assertTrue(verify.verifyNumType('0010', 2))


_run('ValidacionCajaNegra')
```

## 2. Negativos y decimales (solo enteros no negativos)

`verify.verifyNoNegOrDec(n)` devuelve `False` si la cadena tiene signo menos o punto.

| ID | Entrada | Partición | Esperado |
|---|---|---|---|
| CP-01 | `10` | P1 válida | True |
| CP-02 | `0` | P1 válida | True |
| CP-03 | `-5` | P2 inválida | False |
| CP-04 | `1.23` | P3 inválida | False |

```python
class NegativosYDecimalesCajaNegra(unittest.TestCase):

    def test_cp01_entero_positivo(self):
        """CP-01 (P1): '10' se acepta"""
        self.assertTrue(verify.verifyNoNegOrDec('10'))

    def test_cp02_cero(self):
        """CP-02 (P1): '0' se acepta"""
        self.assertTrue(verify.verifyNoNegOrDec('0'))

    def test_cp03_negativo(self):
        """CP-03 (P2): '-5' se rechaza"""
        self.assertFalse(verify.verifyNoNegOrDec('-5'))

    def test_cp04_decimal(self):
        """CP-04 (P3): '1.23' se rechaza"""
        self.assertFalse(verify.verifyNoNegOrDec('1.23'))


_run('NegativosYDecimalesCajaNegra')
```

## 3. Valor decimal (RF-09)

`conversions.toDecimal(n, base)` devuelve `(valor, pasos, Resuestas por pasos)`; comprobamos el valor. Un representante por base.

| ID | Entrada | Base | Esperado |
|---|---|---|---|
| CP-01 | `1010` | 2 | 10 |
| CP-02 | `17` | 8 | 15 |
| CP-03 | `FF` | 16 | 255 |
| CP-04 | `ff` | 16 | 255 (RF-06) |
| CP-05 | `0010` | 2 | 2 (RF-07) |

```python
class ValorDecimalCajaNegra(unittest.TestCase):

    def test_cp01_binario_a_decimal(self):
        """CP-01 (RF-09): 1010(2) == 10"""
        self.assertEqual(conversions.toDecimal('1010', 2)[0], 10)

    def test_cp02_octal_a_decimal(self):
        """CP-02 (RF-09): 17(8) == 15"""
        self.assertEqual(conversions.toDecimal('17', 8)[0], 15)

    def test_cp03_hex_a_decimal_mayuscula(self):
        """CP-03 (RF-09): FF(16) == 255"""
        self.assertEqual(conversions.toDecimal('FF', 16)[0], 255)

    def test_cp04_hex_a_decimal_minuscula(self):
        """CP-04 (RF-06, RF-09): ff(16) == 255"""
        self.assertEqual(conversions.toDecimal('ff', 16)[0], 255)

    def test_cp05_ceros_a_la_izquierda(self):
        """CP-05 (RF-07): 0010(2) == 2"""
        self.assertEqual(conversions.toDecimal('0010', 2)[0], 2)


_run('ValorDecimalCajaNegra')
```

## 4. Decimal a otras bases (RF-10, RF-11, RF-12, RF-14, RF-15)

`conversions.decimalTo(n, base)` devuelve `(digitos, pasos, lista)` por divisiones sucesivas.

- RF-10 / RF-11: valor correcto en bases 2, 8 y 16.
- RF-12: hex en mayúscula, sin ceros a la izquierda.
- RF-14: cada paso con dividendo, cociente y residuo.
- RF-15: cero devuelve `'0'` sin pasos.

```python
class DecimalAOtrasBasesCajaNegra(unittest.TestCase):

    def test_cp01_decimal_a_binario(self):
        """CP-01 (RF-10, RF-11): 10 -> '1010'"""
        self.assertEqual(conversions.decimalTo(10, 2)[0], '1010')

    def test_cp02_decimal_a_octal(self):
        """CP-02 (RF-10, RF-11): 64 -> '100'"""
        self.assertEqual(conversions.decimalTo(64, 8)[0], '100')

    def test_cp03_decimal_a_hex(self):
        """CP-03 (RF-10, RF-11): 255 -> 'FF'"""
        self.assertEqual(conversions.decimalTo(255, 16)[0], 'FF')

    def test_cp04_hex_en_mayuscula(self):
        """CP-04 (RF-12): 2748 -> 'ABC'"""
        self.assertEqual(conversions.decimalTo(2748, 16)[0], 'ABC')

    def test_cp05_hex_sin_ceros_a_la_izquierda(self):
        """CP-05 (RF-12): 16 -> '10', sin cero a la izquierda"""
        result = conversions.decimalTo(16, 16)[0]
        self.assertEqual(result, '10')
        self.assertFalse(result.startswith('0'))

    def test_cp06_pasos_de_division(self):
        """CP-06 (RF-14): los pasos de 10 -> binario listan cada división"""
        steps = conversions.decimalTo(10, 2)[1]
        self.assertEqual(steps[0], '10 / 2 = 5 remainder = 0')
        self.assertEqual(steps[-1], '1 / 2 = 0 remainder = 1')

    def test_cp07_cero_sin_pasos(self):
        """CP-07 (RF-15): 0 -> '0' y lista de pasos vacía"""
        digits, steps, _ = conversions.decimalTo(0, 2)
        self.assertEqual(digits, '0')
        self.assertEqual(steps, [])


_run('DecimalAOtrasBasesCajaNegra')
```

## 5. Boundary value analysis (RF-08, RF-15)

El valor decimal debe estar entre 0 y 1.000.000. La app lo verifica en `index.calculate` con `decimal > 1000000`; probamos los bordes con la misma regla.

| ID | Valor | Borde | ¿Acepta? |
|---|---|---|---|
| CP-01 | 0 | límite inferior (RF-15) | sí |
| CP-02 | 1.000.000 | máximo exacto | sí |
| CP-03 | 1.000.001 | justo por encima | no (RF-08) |
| CP-04 | 999.999 | justo por debajo | sí |

```python
MAX_VALUE = 1000000

def value_is_accepted(decimal):
    return 0 <= decimal <= MAX_VALUE


class RangoLimiteCajaNegra(unittest.TestCase):

    def test_cp01_limite_inferior_cero(self):
        """CP-01 (RF-15): 0 se acepta"""
        self.assertTrue(value_is_accepted(0))

    def test_cp02_limite_superior_exacto(self):
        """CP-02 (RF-08): exactamente 1.000.000 se acepta"""
        self.assertTrue(value_is_accepted(1000000))

    def test_cp03_justo_por_encima(self):
        """CP-03 (RF-08): 1.000.001 se rechaza"""
        self.assertFalse(value_is_accepted(1000001))

    def test_cp04_justo_por_debajo(self):
        """CP-04 (RF-08): 999.999 se acepta"""
        self.assertTrue(value_is_accepted(999999))


_run('RangoLimiteCajaNegra')
```

## 6. Consistencia entre las cuatro bases (RF-09, RF-10, RF-11)

Round-trip: `toDecimal` y de vuelta con `decimalTo`. Un representante por base de origen; las cuatro representaciones deben coincidir en el mismo valor.

| ID | Origen | Base | Decimal | Binario | Octal | Hex |
|---|---|---|---|---|---|---|
| CP-01 | `1010` | 2 | 10 | 1010 | 12 | A |
| CP-02 | `17` | 8 | 15 | 1111 | 17 | F |
| CP-03 | `FF` | 16 | 255 | 11111111 | 377 | FF |

```python
class ConsistenciaBasesCajaNegra(unittest.TestCase):

    def _check(self, text, base, dec, binv, octv, hexv):
        decimal = conversions.toDecimal(text, base)[0] if base != 10 else int(text)
        self.assertEqual(decimal, dec)
        self.assertEqual(conversions.decimalTo(decimal, 2)[0], binv)
        self.assertEqual(conversions.decimalTo(decimal, 8)[0], octv)
        self.assertEqual(conversions.decimalTo(decimal, 16)[0], hexv)

    def test_cp01_desde_binario(self):
        """CP-01 (RF-11): 1010(2) en las cuatro bases"""
        self._check('1010', 2, 10, '1010', '12', 'A')

    def test_cp02_desde_octal(self):
        """CP-02 (RF-11): 17(8) en las cuatro bases"""
        self._check('17', 8, 15, '1111', '17', 'F')

    def test_cp03_desde_hex(self):
        """CP-03 (RF-11): FF(16) en las cuatro bases"""
        self._check('FF', 16, 255, '11111111', '377', 'FF')


_run('ConsistenciaBasesCajaNegra')
```
